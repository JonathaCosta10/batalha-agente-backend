###  Visão Geral do Projeto
O **i.agora** é um agente financeiro inteligente desenvolvido para transformar intenções e dados transacionais em futuros simulados e comparáveis, ajudando correntistas a tomarem decisões com contexto, clareza e antecipação de impactos.

Diferente dos aplicativos bancários tradicionais que olham apenas para o saldo atual ou para a parcela "que cabe no mês", o **i.agora** atua em três pilares fundamentais:
1. **Entender:** Interpreta o histórico de transações, padrões de renda, recorrências e sazonalidades.
2. **Antecipar:** Projeta o impacto financeiro de decisões de curto e longo prazo nos próximos 6 meses.
3. **Orientar:** Apresenta cenários, *trade-offs* e alternativas personalizadas, conectando o correntista aos melhores produtos do ecossistema Itaú respaldados por aterramento factual via RAG.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. SIMULAÇÃO DE DADOS GRANULARES (AUTÔNOMOS VS CLT)
# ==========================================
np.random.seed(42)
n_amostra = 15000

# Separando a população: 65% Autônomos (Recebimentos Diversos) e 35% CLT (Salário Fixo)
tipo_perfil = np.random.choice(
    ['Autônomos (Recebimentos Diversos)', 'CLT (Salário Fixo)'],
    size=n_amostra,
    p=[0.65, 0.35]
)

# Gerando rendas mensais com comportamentos distintos (Autônomos com maior dispersão/volatilidade)
rendas = []
for perfil in tipo_perfil:
    if perfil == 'Autônomos (Recebimentos Diversos)':
        rendas.append(np.random.normal(3100, 1600))
    else:
        rendas.append(np.random.normal(4800, 1100))

df_perfil = pd.DataFrame({
    'tipo_perfil': tipo_perfil,
    'renda_mensal': np.clip(rendas, 1200, 22000)
})

# ==========================================
# 2. DEFINIÇÃO DA FAIXA SALARIAL
# ==========================================
def classificar_faixa_salarial(valor):
    if valor <= 2500:
        return 'Até R$ 2.500'
    elif valor <= 5000:
        return 'R$ 2.501 a R$ 5.000'
    elif valor <= 10000:
        return 'R$ 5.001 a R$ 10.000'
    else:
        return 'Acima de R$ 10.000'

df_perfil['faixa_salarial'] = df_perfil['renda_mensal'].apply(classificar_faixa_salarial)
ordem_faixas = ['Até R$ 2.500', 'R$ 2.501 a R$ 5.000', 'R$ 5.001 a R$ 10.000', 'Acima de R$ 10.000']

# Tabela cruzada percentual para análise
tabela_cruzada = pd.crosstab(
    df_perfil['faixa_salarial'],
    df_perfil['tipo_perfil'],
    normalize='columns'
) * 100

# Reindexando para garantir a ordem correta e preenchendo NaNs com 0
tabela_cruzada = tabela_cruzada.reindex(ordem_faixas).fillna(0)

# Formatando visualmente com duas casas decimais e o sinal de porcentagem (%)
tabela_formatada = tabela_cruzada.round(2).astype(str) + '%'

print("--- TABELA: DISTRIBUIÇÃO PERCENTUAL POR FAIXA SALARIAL (%) ---")
display(tabela_formatada)

# ==========================================
# 3. VISUALIZAÇÃO GRÁFICA (2 GRÁFICOS COMPARATIVOS)
# ==========================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Cores estratégicas: Laranja (Autônomos / Diversos), Azul Itaú (CLT / Fixo)
cores_perfil = {
    'Autônomos (Recebimentos Diversos)': '#EC7000',
    'CLT (Salário Fixo)': '#0046B5'
}

# --- Gráfico 1: Curva de Densidade de Renda (Mostrando a Volatilidade dos Autônomos) ---
sns.kdeplot(
    data=df_perfil,
    x='renda_mensal',
    hue='tipo_perfil',
    palette=cores_perfil,
    fill=True,
    common_norm=False,
    alpha=0.4,
    ax=axes[0]
)
axes[0].set_title('Perfil de Renda: Autônomos vs. CLT\n(Evidenciando a dispersão e imprevisibilidade dos autônomos)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Renda Mensal (R$)', fontsize=12)
axes[0].set_ylabel('Densidade Populacional', fontsize=12)
axes[0].set_xlim(1000, 16000)

# --- Gráfico 2: Quantidade de Clientes agrupados por Faixa Salarial e Tipo de Renda ---
sns.countplot(
    data=df_perfil,
    x='faixa_salarial',
    hue='tipo_perfil',
    order=ordem_faixas,
    palette=cores_perfil,
    ax=axes[1]
)
axes[1].set_title('Concentração de Clientes por Faixa Salarial e Categoria', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Faixa Salarial', fontsize=12)
axes[1].set_ylabel('Número de Clientes', fontsize=12)
axes[1].tick_params(axis='x', rotation=15)
axes[1].legend(title='Tipo de Vínculo')

plt.tight_layout()
plt.show()
```

### Análise dos Resultados: Perfil de Renda (Autônomos vs. CLT)

* **Evidência:** O gráfico de densidade (KDE Plot) demonstra que a renda dos **Autônomos (Recebimentos Diversos)** possui uma curva muito mais achatada e ampla, com o pico centrado em torno de **R$ 3.100**.
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. GERANDO DADOS SINTÉTICOS (50.000 clientes)
# ==========================================
np.random.seed(42)
n_clientes = 50000

# Renda Mensal
renda_mensal = np.random.lognormal(mean=8, sigma=0.8, size=n_clientes)
renda_mensal = np.clip(renda_mensal, 1500, 30000)

# Gastos Totais
fator_gasto = np.random.normal(loc=0.95, scale=0.25, size=n_clientes)
gasto_mensal = renda_mensal * fator_gasto

df_clientes = pd.DataFrame({
    'id_cliente': [f'CLT_{i:05d}' for i in range(n_clientes)],
    'renda_mensal': renda_mensal,
    'gasto_mensal': gasto_mensal
})

# ==========================================
# 2. CLASSIFICAÇÃO DOS 3 GRUPOS
# ==========================================
df_clientes['saldo_final'] = df_clientes['renda_mensal'] - df_clientes['gasto_mensal']
df_clientes['percentual_sobra'] = (df_clientes['saldo_final'] / df_clientes['renda_mensal']) * 100

def classificar_status(row):
    if row['saldo_final'] < 0:
        return '1. Endividado (< 0)'
    elif row['percentual_sobra'] < 15:
        return '2. Vulnerável (< 15%)'
    else:
        return '3. Saudável (>= 15%)'

df_clientes['status_fim_mes'] = df_clientes.apply(classificar_status, axis=1)

# ==========================================
# 3. SIMULANDO O COMPORTAMENTO DE GASTO POR CATEGORIA
# ==========================================
# Vamos distribuir o 'gasto_mensal' em 4 categorias dependendo do grupo do cliente
dados_categorias = []

for status in df_clientes['status_fim_mes'].unique():
    grupo = df_clientes[df_clientes['status_fim_mes'] == status].copy()
    n_group = len(grupo)

    if 'Endividado' in status:
        # Alto custo com juros/taxas e gastos fixos
        casa = np.random.normal(0.40, 0.05, n_group)
        mercado = np.random.normal(0.20, 0.05, n_group)
        juros = np.random.normal(0.25, 0.05, n_group)
        lazer = np.random.normal(0.15, 0.05, n_group)
    elif 'Vulnerável' in status:
        # Alto custo discricionário (Lazer) que corrói a margem
        casa = np.random.normal(0.35, 0.05, n_group)
        mercado = np.random.normal(0.20, 0.05, n_group)
        juros = np.random.normal(0.10, 0.05, n_group)
        lazer = np.random.normal(0.35, 0.05, n_group)
    else:
        # Saudável
        casa = np.random.normal(0.30, 0.05, n_group)
        mercado = np.random.normal(0.15, 0.05, n_group)
        juros = np.random.normal(0.05, 0.02, n_group)
        lazer = np.random.normal(0.50, 0.05, n_group)

    # Aplicando os percentuais e garantindo que não sejam negativos
    grupo['Casa'] = np.clip(casa, 0.05, 1.0) * grupo['gasto_mensal']
    grupo['Mercado/Essencial'] = np.clip(mercado, 0.05, 1.0) * grupo['gasto_mensal']
    grupo['Produtos Financeiros (Juros)'] = np.clip(juros, 0.0, 1.0) * grupo['gasto_mensal']
    grupo['Lazer/Discricionário'] = np.clip(lazer, 0.0, 1.0) * grupo['gasto_mensal']

    # Normalizando para bater exatamente com o gasto_mensal total
    somas = grupo[['Casa', 'Mercado/Essencial', 'Produtos Financeiros (Juros)', 'Lazer/Discricionário']].sum(axis=1)
    for col in ['Casa', 'Mercado/Essencial', 'Produtos Financeiros (Juros)', 'Lazer/Discricionário']:
        grupo[col] = (grupo[col] / somas) * grupo['gasto_mensal']

    dados_categorias.append(grupo)

df_completo = pd.concat(dados_categorias)

# Preparando dados para o gráfico de barras empilhadas (Média de percentual de gasto por categoria)
colunas_gasto = ['Casa', 'Mercado/Essencial', 'Produtos Financeiros (Juros)', 'Lazer/Discricionário']
resumo_gastos = df_completo.groupby('status_fim_mes')[colunas_gasto].mean()
resumo_gastos_pct = resumo_gastos.div(resumo_gastos.sum(axis=1), axis=0) * 100

# ==========================================
# 4. VISUALIZAÇÃO DOS RESULTADOS (2 GRÁFICOS)
# ==========================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# --- Gráfico 1: Tamanho da População ---
percentual_status = df_clientes['status_fim_mes'].value_counts(normalize=True).sort_index() * 100
cores_pop = ['#d62728', '#ec7000', '#0046b5'] # Vermelho, Laranja, Azul Itaú

sns.barplot(x=percentual_status.values, y=percentual_status.index, palette=cores_pop, ax=axes[0])
axes[0].set_title('Distribuição da População (A Dor)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Percentual da Base (%)')
axes[0].set_ylabel('')
axes[0].set_xlim(0, 50)

for p in axes[0].patches:
    width = p.get_width()
    axes[0].text(width + 1, p.get_y() + p.get_height()/2., f'{width:.1f}%',
             ha="left", va="center", fontweight='bold')

# --- Gráfico 2: Como eles gastam o dinheiro (Barras Empilhadas) ---
cores_cat = ['#5e5e5e', '#a6a6a6', '#d62728', '#ffb84d'] # Casa, Mercado, Juros (Vermelho), Lazer (Amarelo)
resumo_gastos_pct.plot(kind='barh', stacked=True, color=cores_cat, ax=axes[1], width=0.7)

axes[1].set_title('Como o dinheiro é gasto? (Composição das Saídas)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Percentual do Gasto Total (%)')
axes[1].set_ylabel('')
axes[1].legend(title='Categorias', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adicionando rótulos dentro das barras
for c in axes[1].containers:
    axes[1].bar_label(c, fmt='%.1f%%', label_type='center', color='white', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. SIMULAÇÃO DE DADOS DE GASTO E PRODUTOS DO BANCO POR PERFIL
# ==========================================
np.random.seed(42)
n_amostra = 15000

perfis = ['1. Endividado (< 0)', '2. Vulnerável (< 15%)', '3. Saudável (>= 15%)']
dados_banco = []

for perfil in perfis:
    if 'Endividado' in perfil:
        gasto_banco_pct = np.random.normal(24.5, 4.0, n_amostra)
        produto_sugerido = 'Crédito Consolidado com Juros Reduzidos / Renegociação IA'
        receita_media_banco = 320.0
    elif 'Vulnerável' in perfil:
        gasto_banco_pct = np.random.normal(12.0, 2.5, n_amostra)
        produto_sugerido = 'Reserva Automática / Guardar Dinheiro com Rendimento'
        receita_media_banco = 110.0
    else:
        gasto_banco_pct = np.random.normal(3.5, 1.2, n_amostra)
        produto_sugerido = 'CDB Itaú Diário / Fundos de Investimento / Previdência'
        receita_media_banco = 45.0

    for i in range(n_amostra):
        dados_banco.append({
            'perfil_cliente': perfil,
            'gasto_banco_pct': max(0.5, gasto_banco_pct[i]),
            'produto_recomendado': produto_sugerido,
            'receita_banco_rs': receita_media_banco  # Corrigido para evitar caracteres inválidos
        })

df_banco = pd.DataFrame(dados_banco)

# ==========================================
# 2. TABELA CONSOLIDADA (PRODUTOS, GASTO E RETORNO)
# ==========================================
# Corrigido o nome do argumento nomeado (removido o caractere '$' que causava o erro de sintaxe)
tabela_cruzada = df_banco.groupby('perfil_cliente').agg(
    Gasto_Medio_Com_Banco_Pct=('gasto_banco_pct', 'mean'),
    Receita_Media_Banco_RS=('receita_banco_rs', 'mean'),
    Melhor_Produto_Para_Perfil=('produto_recomendado', 'first')
).reset_index()

# Ajustando nome das colunas para exibição limpa
tabela_cruzada.columns = ['Perfil do Cliente', '% Gasto Médio com o Banco', 'Receita Média do Banco (R$)', 'Melhor Produto para o Perfil (Bem-Estar)']
tabela_cruzada['% Gasto Médio com o Banco'] = tabela_cruzada['% Gasto Médio com o Banco'].round(2).astype(str) + '%'
tabela_cruzada['Receita Média do Banco (R$)'] = 'R$ ' + tabela_cruzada['Receita Média do Banco (R$)'].round(2).astype(str)

print("--- MATRIZ DE PRODUTOS E MONETIZAÇÃO POR PERFIL ---")
display(tabela_cruzada)

# ==========================================
# 3. VISUALIZAÇÃO: COMPARAÇÃO DO COMPROMETIMENTO DA RENDA COM O BANCO
# ==========================================
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

cores = ['#d62728', '#ec7000', '#0046b5']
ax = sns.barplot(
    data=df_banco,
    x='gasto_banco_pct',
    y='perfil_cliente',
    palette=cores,
    orient='h'
)

plt.title('Impacto Financeiro: Quanto da Renda do Cliente é Retida pelo Banco\n(Juros, Tarifas e Produtos)', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Percentual da Renda Gasto com Taxas/Serviços Bancários (%)', fontsize=12)
plt.ylabel('')
plt.xlim(0, 35)

for p in ax.patches:
    width = p.get_width()
    ax.text(width + 0.8, p.get_y() + p.get_height()/2., f'{width:.1f}%',
            ha="left", va="center", fontweight='bold', fontsize=11)

plt.tight_layout()
plt.show()
```

* *Insight:* O endividado está preso numa espiral onde uma fatia massiva da renda serve apenas para pagar o próprio custo da dívida (rotativo e cheque especial), sufocando o orçamento básico.

Esta segmentação cruza a volumetria do problema com a causa raiz do comportamento de consumo. O agente **i.agora** deixa de ser apenas um visualizador de extratos e passa a atuar cirurgicamente:

1. **Para os Endividados (42,2%):** Intercepta o peso insustentável dos juros (25%) oferecendo renegociações inteligentes via RAG.


2. **Para os Vulneráveis (23,4%):** Atua preventivamente para evitar que escorreguem para o vermelho, otimizando margens estreitas.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ==============================================================================
# ANÁLISE COMPORTAMENTAL: A REGRA 50/30/20 E PROJEÇÃO DE CORTES
# ==============================================================================
print("--- PROCESSANDO SEGMENTAÇÃO ORÇAMENTÁRIA (50/30/20) ---")

# 1. Simulando ou utilizando a base de clientes existente com categorias mapeadas
np.random.seed(42)
n_amostra = 5000

df_orcamento = pd.DataFrame({
    'id_usuario': [f'CLT_{i:04d}' for i in range(n_amostra)],
    'renda_mensal': np.random.normal(4500, 1200, n_amostra).clip(1500, 20000)
})

# 2. Definindo a taxonomia dos blocos da regra 50/30/20:
# - 50% Gastos Fixos (Moradia, Contas Essenciais, Saúde, Educação)
# - 30% Gastos Variáveis / Estilo de Vida (Delivery, Lazer, Compras, Assinaturas)
# - 20% Investimentos / Poupança / Reserva (ou abatimento de dívidas)
df_orcamento['gasto_fixo'] = df_orcamento['renda_mensal'] * np.random.uniform(0.35, 0.65, n_amostra)       # Alvo: <= 50%
df_orcamento['gasto_variavel'] = df_orcamento['renda_mensal'] * np.random.uniform(0.15, 0.45, n_amostra) # Alvo: <= 30%
df_orcamento['investimento_poupanca'] = df_orcamento['renda_mensal'] - (df_orcamento['gasto_fixo'] + df_orcamento['gasto_variavel'])

# Calculando os percentuais reais em relação à renda
df_orcamento['pct_fixo'] = (df_orcamento['gasto_fixo'] / df_orcamento['renda_mensal']) * 100
df_orcamento['pct_variavel'] = (df_orcamento['gasto_variavel'] / df_orcamento['renda_mensal']) * 100
df_orcamento['pct_investimento'] = (df_orcamento['investimento_poupanca'] / df_orcamento['renda_mensal']) * 100

# 3. Lógica de Diagnóstico e Sugestão de Próximo Corte pelo Agente
def gerar_sugestao_corte(row):
    if row['pct_fixo'] > 53:
        return 'Alerta Fixo: Custo de moradia/essenciais acima de 50%. Avaliar renegociação de contratos.'
    elif row['pct_variavel'] > 33:
        return 'Oportunidade Variável: Delivery e Lazer consomem mais de 30%. Sugerir meta de corte de 15% nestas categorias.'
    elif row['pct_investimento'] < 10:
        return 'Construção de Futuro: Poupança abaixo de 20%. Sugerir automatizar aplicação de R$ 50 após o recebimento.'
    else:
        return 'Perfil Equilibrado: Orçamento alinhado à regra 50/30/20. Manter estratégia de investimentos.'

df_orcamento['proximo_corte_sugerido'] = df_orcamento.apply(gerar_sugestao_corte, axis=1)

# Exibindo amostra da inteligência do agente aplicada ao orçamento
print("\n--- AMOSTRA DE DIAGNÓSTICO E SUGESTÃO DE AJUSTE ---")
display(df_orcamento[['id_usuario', 'renda_mensal', 'pct_fixo', 'pct_variavel', 'pct_investimento', 'proximo_corte_sugerido']].head(8))

# ==============================================================================
# 4. VISUALIZAÇÃO GRÁFICA: DESVIO DO GASTO VARIÁVEL VS META (30%)
# ==============================================================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Distribuição dos Gastos Variáveis em relação à Meta de 30%
sns.histplot(df_orcamento['pct_variavel'], color='#EC7000', kde=True, ax=axes[0])
axes[0].axvline(30, color='#0046b5', linestyle='--', linewidth=2.5, label='Meta Ideal Regra 50/30/20 (30%)')
axes[0].set_title('Desvio do Gasto Variável (Lazer/Delivery)\n(Onde o agente identifica margem para cortes indolores)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Comprometimento da Renda com Gastos Variáveis (%)', fontsize=11)
axes[0].set_ylabel('Número de Correntistas', fontsize=11)
axes[0].legend(fontsize=11)

# Gráfico 2: Contagem dos Tipos de Alerta / Próximo Corte Gerado pelo Agente
ordem_alertas = df_orcamento['proximo_corte_sugerido'].value_counts().index
sns.countplot(data=df_orcamento, y='proximo_corte_sugerido', order=ordem_alertas, palette='Blues_r', ax=axes[1])
axes[1].set_title('Ações Recomendadas pelo Agente i.agora\n(Segmentação para Projeção Futura)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Quantidade de Clientes', fontsize=11)
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# ==============================================================================
# 1. BUSCA AUTOMÁTICA DO ARQUIVO CSV NO DIRETÓRIO DO PROJETO
# ==============================================================================
diretorio_base = r"C:\Users\lostj\Documents\Data\projects\batalha de agentes"
padrao_busca = os.path.join(diretorio_base, "**", "*bq-results*.csv")
arquivos_encontrados = glob.glob(padrao_busca, recursive=True)

df_agg = pd.DataFrame()

if arquivos_encontrados:
    arquivo_alvo = arquivos_encontrados[0]
    print(f"Sucesso: Arquivo CSV localizado automaticamente em:\n -> {arquivo_alvo}")
    try:
        df_agg = pd.read_csv(arquivo_alvo, sep=",")
        if "Table name" in str(df_agg.columns[0]):
            df_agg = pd.read_csv(arquivo_alvo, sep=",", skiprows=1)
        df_agg.columns = df_agg.columns.str.strip().str.lower()
    except Exception as e:
        print(f"Aviso ao ler o arquivo CSV: {e}. Ativando fallback...")
        df_agg = pd.DataFrame()

if df_agg.empty:
    print("Aviso: Ativando fallback com matriz de categorias simuladas...")
    dados_fallback = [
        {"tipo": "E", "nom_cate_macro": "Salarios e bonificacoes"},
        {"tipo": "E", "nom_cate_macro": "Recebimentos diversos"},
        {"tipo": "S", "nom_cate_macro": "Casa"},
        {"tipo": "S", "nom_cate_macro": "Mercado"},
        {"tipo": "S", "nom_cate_macro": "Restaurantes"},
        {"tipo": "S", "nom_cate_macro": "Lazer e Imprevistos"},
        {"tipo": "S", "nom_cate_macro": "Produtos financeiros"},
    ]
    df_agg = pd.DataFrame(dados_fallback)

# ==============================================================================
# 2. SIMULAÇÃO TRANSACIONAL COM DÉFICIT EM DEZEMBRO
# ==============================================================================
print("Processando base e gerando histórico transacional (Jan-Dez 2025)...")
np.random.seed(42)
n_clientes = 2000
usuarios = [f"CLT_{i:04d}" for i in range(n_clientes)]

transacoes = []
for mes in range(1, 13):
    for u in usuarios:
        perfil_risco = int(u.split("_")[1]) < (n_clientes * 0.35)

        for _, row in df_agg.iterrows():
            tipo = str(row["tipo"]).upper()
            macro = str(row["nom_cate_macro"])

            if tipo == "E":
                # Entradas estáveis de renda mensal
                vlr = np.random.uniform(1800, 3200)
            else:
                vlr = np.random.uniform(150, 500)
                # Multiplicador acentuado de despesas para perfil de risco no fim de ano
                if perfil_risco and mes >= 11:
                    vlr = vlr * np.random.uniform(2.2, 4.0)

            transacoes.append({
                "id_usuario": u,
                "mes": mes,
                "tipo": tipo,
                "nom_cate_macro": macro,
                "valor_direcional": vlr if tipo == "E" else -vlr,
            })

df = pd.DataFrame(transacoes)

# ==============================================================================
# 3. ISOLAMENTO DA POPULAÇÃO NEGATIVADA EM DEZEMBRO
# ==============================================================================
df_dez = df[df["mes"] == 12]
saldo_dezembro = (
    df_dez.groupby("id_usuario")["valor_direcional"].sum().reset_index()
)
saldo_dezembro.rename(
    columns={"valor_direcional": "divida_atual"}, inplace=True
)

negativados_dez = saldo_dezembro[saldo_dezembro["divida_atual"] < 0].copy()
ids_alvo = negativados_dez["id_usuario"].unique()

print(f"Total de clientes na base: {n_clientes}")
print(f"População isolada (Negativados em Dezembro): {len(ids_alvo)} clientes")

if len(ids_alvo) > 0:
    df_coorte = df[df["id_usuario"].isin(ids_alvo)].copy()

    # ==============================================================================
    # 4. MATRIZ DE ESSENCIALIDADE E PLANO DE MITIGAÇÃO
    # ==============================================================================
    def classificar_essencialidade(categoria):
        cat_lower = str(categoria).lower()
        if any(x in cat_lower for x in ["casa", "mercado", "saúde", "educacao"]):
            return "Nível 3 (Essencial - Não Cortar)"
        elif any(
            x in cat_lower for x in ["restaurante", "cuidados", "transporte"]
        ):
            return "Nível 2 (Flexível - Reduzir)"
        else:
            return "Nível 1 (Supérfluo/Lazer - Cortar Primeiro)"

    df_coorte["prioridade_corte"] = df_coorte["nom_cate_macro"].apply(
        classificar_essencialidade
    )

    gastos_dez_alvo = df_coorte[
        (df_coorte["mes"] == 12) & (df_coorte["tipo"] == "S")
    ].copy()
    gastos_dez_alvo["valor_gasto"] = gastos_dez_alvo["valor_direcional"].abs()

    perfil_mitigacao = (
        gastos_dez_alvo.groupby(["id_usuario", "prioridade_corte"])[
            "valor_gasto"
        ]
        .sum()
        .unstack(fill_value=0)
        .reset_index()
    )

    for col in [
        "Nível 1 (Supérfluo/Lazer - Cortar Primeiro)",
        "Nível 2 (Flexível - Reduzir)",
        "Nível 3 (Essencial - Não Cortar)",
    ]:
        if col not in perfil_mitigacao.columns:
            perfil_mitigacao[col] = 0.0

    perfil_mitigacao = pd.merge(
        perfil_mitigacao, negativados_dez, on="id_usuario"
    )
    perfil_mitigacao["divida_atual"] = perfil_mitigacao["divida_atual"].abs()

    perfil_mitigacao["margem_corte_seguro"] = perfil_mitigacao[
        "Nível 1 (Supérfluo/Lazer - Cortar Primeiro)"
    ] + (perfil_mitigacao["Nível 2 (Flexível - Reduzir)"] * 0.5)

    perfil_mitigacao["plano_possivel_sem_afetar_essencial"] = (
        perfil_mitigacao["margem_corte_seguro"]
        >= perfil_mitigacao["divida_atual"]
    )

    print("\n--- PLANO DE MITIGAÇÃO: AMOSTRA DE CLIENTES ---")
    display(
        perfil_mitigacao[[
            "id_usuario",
            "divida_atual",
            "margem_corte_seguro",
            "plano_possivel_sem_afetar_essencial",
        ]].head()
    )

    # ==============================================================================
    # 5. VISUALIZAÇÃO: EVOLUÇÃO TEMPORAL DA COORTE
    # ==============================================================================
    evolucao_coorte = (
        df_coorte.groupby("mes")
        .agg(
            entradas_medias=(
                "valor_direcional",
                lambda x: x[x > 0].sum() / len(ids_alvo),
            ),
            saidas_medias=(
                "valor_direcional",
                lambda x: x[x < 0].abs().sum() / len(ids_alvo),
            ),
        )
        .reset_index()
    )

    meses_nome = {
        1: "Jan",
        2: "Fev",
        3: "Mar",
        4: "Abr",
        5: "Mai",
        6: "Jun",
        7: "Jul",
        8: "Ago",
        9: "Set",
        10: "Out",
        11: "Nov",
        12: "Dez",
    }
    evolucao_coorte["nome_mes"] = evolucao_coorte["mes"].map(meses_nome)

    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(12, 6))

    plt.plot(
        evolucao_coorte["nome_mes"],
        evolucao_coorte["entradas_medias"],
        label="Renda Média (Entradas)",
        color="#0046b5",
        marker="o",
        linewidth=2.5,
    )
    plt.plot(
        evolucao_coorte["nome_mes"],
        evolucao_coorte["saidas_medias"],
        label="Gastos Médios (Saídas)",
        color="#d62728",
        marker="s",
        linewidth=2.5,
    )

    plt.axhline(
        evolucao_coorte["entradas_medias"].mean(),
        color="gray",
        linestyle="--",
        alpha=0.5,
        label="Média Anual de Renda",
    )

    plt.title(
        "Evolução do Descontrole: População Negativada em Dezembro",
        fontsize=14,
        fontweight="bold",
        pad=15,
    )
    plt.xlabel("Meses de 2025", fontsize=12)
    plt.ylabel("Volume Financeiro Médio (R$)", fontsize=12)
    plt.legend(fontsize=11)

    plt.tight_layout()
    plt.show()
else:
    print(
        "Nenhum cliente negativado encontrado para gerar as métricas do gráfico."
    )
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# ==============================================================================
# 1. LEITURA DA BASE AGREGADA E EXPANSÃO TRANSACIONAL (Para criar datas e IDs)
# ==============================================================================
arquivo_alvo = r"C:\Users\lostj\Documents\Data\projects\batalha de agentes\.ipynb_checkpoints\bq-results-20260926-143453-1790433308679 - bq-results-20260926-143453-1790433308679.csv.csv"

try:
    df_agg = pd.read_csv(arquivo_alvo, sep=',')
    if 'Table name' in str(df_agg.columns[0]):
        df_agg = pd.read_csv(arquivo_alvo, sep=',', skiprows=1)

    df_agg.columns = df_agg.columns.str.strip().str.lower()

    print("Processando base e gerando histórico transacional (Jan-Dez 2025)...")
    np.random.seed(42)
    n_clientes = 2000
    usuarios = [f'CLT_{i:04d}' for i in range(n_clientes)]

    transacoes = []
    # Simulando um ano de transações baseadas nas categorias do seu arquivo
    for mes in range(1, 13):
        for u in usuarios:
            # Forçando 30% da base a se descontrolar no fim do ano (nossa coorte alvo)
            perfil_risco = int(u.split('_')[1]) < (n_clientes * 0.3)

            for _, row in df_agg.iterrows():
                tipo = row['tipo'].upper()
                macro = str(row['nom_cate_macro'])

                # Definindo valores base
                if tipo == 'E':
                    vlr = np.random.uniform(1500, 5000) if mes != 12 else np.random.uniform(1500, 6000)
                else:
                    vlr = np.random.uniform(50, 600)
                    # Se for perfil de risco, os gastos explodem em Nov/Dez
                    if perfil_risco and mes >= 11:
                        vlr = vlr * np.random.uniform(1.5, 2.5)

                transacoes.append({
                    'id_usuario': u,
                    'mes': mes,
                    'tipo': tipo,
                    'nom_cate_macro': macro,
                    'valor_direcional': vlr if tipo == 'E' else -vlr
                })

    df = pd.DataFrame(transacoes)

    # ==============================================================================
    # 2. ISOLAR A POPULAÇÃO NEGATIVADA EM DEZEMBRO
    # ==============================================================================
    df_dez = df[df['mes'] == 12]
    saldo_dezembro = df_dez.groupby('id_usuario')['valor_direcional'].sum().reset_index()
    saldo_dezembro.rename(columns={'valor_direcional': 'divida_atual'}, inplace=True)

    # Identifica quem fechou Dezembro < 0
    negativados_dez = saldo_dezembro[saldo_dezembro['divida_atual'] < 0]
    ids_alvo = negativados_dez['id_usuario'].unique()

    print(f"Total de clientes na base: {n_clientes}")
    print(f"População isolada (Negativados em Dezembro): {len(ids_alvo)} clientes")

    # Histórico Jan-Dez APENAS dessa população
    df_coorte = df[df['id_usuario'].isin(ids_alvo)].copy()

    # ==============================================================================
    # 3. CRIAÇÃO DA NOVA VARIÁVEL E PLANO DE MITIGAÇÃO
    # ==============================================================================
    # Matriz de Essencialidade (O que cortar primeiro)
    def classificar_essencialidade(categoria):
        cat_lower = str(categoria).lower()
        if any(x in cat_lower for x in ['casa', 'mercado', 'saúde', 'educacao']):
            return 'Nível 3 (Essencial - Não Cortar)'
        elif any(x in cat_lower for x in ['restaurante', 'cuidados', 'transporte']):
            return 'Nível 2 (Flexível - Reduzir)'
        else:
            return 'Nível 1 (Supérfluo/Lazer - Cortar Primeiro)'

    df_coorte['prioridade_corte'] = df_coorte['nom_cate_macro'].apply(classificar_essencialidade)

    # Pegar apenas os gastos (saídas) de dezembro dessa população
    gastos_dez_alvo = df_coorte[(df_coorte['mes'] == 12) & (df_coorte['tipo'] == 'S')].copy()
    gastos_dez_alvo['valor_gasto'] = gastos_dez_alvo['valor_direcional'].abs()

    # Agrupando por cliente e prioridade de corte
    perfil_mitigacao = gastos_dez_alvo.groupby(['id_usuario', 'prioridade_corte'])['valor_gasto'].sum().unstack(fill_value=0).reset_index()

    # Juntando com a dívida do momento do contato
    perfil_mitigacao = pd.merge(perfil_mitigacao, negativados_dez, on='id_usuario')
    perfil_mitigacao['divida_atual'] = perfil_mitigacao['divida_atual'].abs() # Tornando positivo para comparar

    # CRIAÇÃO DA NOVA VARIÁVEL: Margem de Corte Seguro
    # Assume-se que o cliente pode cortar 100% do Nível 1 e 50% do Nível 2. O Nível 3 fica intacto.
    if 'Nível 1 (Supérfluo/Lazer - Cortar Primeiro)' in perfil_mitigacao.columns and 'Nível 2 (Flexível - Reduzir)' in perfil_mitigacao.columns:
        perfil_mitigacao['margem_corte_seguro'] = perfil_mitigacao['Nível 1 (Supérfluo/Lazer - Cortar Primeiro)'] + (perfil_mitigacao['Nível 2 (Flexível - Reduzir)'] * 0.5)
    else:
        perfil_mitigacao['margem_corte_seguro'] = 0

    perfil_mitigacao['plano_possivel_sem_afetar_essencial'] = perfil_mitigacao['margem_corte_seguro'] >= perfil_mitigacao['divida_atual']

    print("\n--- PLANO DE MITIGAÇÃO: AMOSTRA DE CLIENTES ---")
    display(perfil_mitigacao[['id_usuario', 'divida_atual', 'margem_corte_seguro', 'plano_possivel_sem_afetar_essencial']].head())

    # ==============================================================================
    # 4. VISUALIZAÇÃO: EVOLUÇÃO ATÉ A DÍVIDA
    # ==============================================================================
    evolucao_coorte = df_coorte.groupby('mes').agg(
        entradas_medias=('valor_direcional', lambda x: x[x > 0].sum() / len(ids_alvo)),
        saidas_medias=('valor_direcional', lambda x: x[x < 0].abs().sum() / len(ids_alvo))
    ).reset_index()

    meses_nome = {1:'Jan', 2:'Fev', 3:'Mar', 4:'Abr', 5:'Mai', 6:'Jun', 7:'Jul', 8:'Ago', 9:'Set', 10:'Out', 11:'Nov', 12:'Dez'}
    evolucao_coorte['nome_mes'] = evolucao_coorte['mes'].map(meses_nome)

    sns.set_theme(style="whitegrid")
    plt.figure(figsize=(12, 6))

    plt.plot(evolucao_coorte['nome_mes'], evolucao_coorte['entradas_medias'], label='Renda Média (Entradas)', color='#0046b5', marker='o', linewidth=2.5)
    plt.plot(evolucao_coorte['nome_mes'], evolucao_coorte['saidas_medias'], label='Gastos Médios (Saídas)', color='#d62728', marker='s', linewidth=2.5)

    plt.axhline(evolucao_coorte['entradas_medias'].mean(), color='gray', linestyle='--', alpha=0.5, label='Média Anual de Renda')

    plt.title('Evolução do Descontrole: População Negativada em Dezembro', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Meses de 2025', fontsize=12)
    plt.ylabel('Volume Financeiro Médio (R$)', fontsize=12)
    plt.legend(fontsize=11)

    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"\n[Erro Crítico]: {e}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ==============================================================================
# 1. GERAÇÃO/EXPANSÃO TRANSACIONAL (Garante que o código rode perfeitamente)
# ==============================================================================
print("Iniciando processamento da malha de dados transacionais...")
np.random.seed(42)
n_clientes = 2500
usuarios = [f'CLT_{i:04d}' for i in range(n_clientes)]

transacoes = []
for mes in range(1, 13):
    for u in usuarios:
        perfil_risco = int(u.split('_')[1]) < (n_clientes * 0.35) # 35% da base tem tendência ao descontrole
        is_clt = int(u.split('_')[1]) % 2 == 0 # Metade é CLT, metade Autônomo

        # --- ENTRADAS (RENDA) ---
        if is_clt:
            transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'E', 'macro': 'Salarios', 'micro': 'Salario CLT', 'valor': np.random.normal(4500, 800)})
            if mes in [11, 12]: # Sazonalidade
                transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'E', 'macro': 'Salarios', 'micro': '13o salario', 'valor': np.random.normal(4500, 800)})
        else:
            transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'E', 'macro': 'Recebimentos diversos', 'micro': 'Recebimentos diversos', 'valor': np.random.normal(3800, 1500)}) # Maior volatilidade

        # --- SAÍDAS (GASTOS) ---
        renda_base = 4500 if is_clt else 3800
        fator_gasto = np.random.uniform(0.7, 1.2) if perfil_risco and mes >= 10 else np.random.uniform(0.6, 0.9)
        gasto_total = renda_base * fator_gasto

        # Distribuição de gastos
        transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'S', 'macro': 'Casa/Essencial', 'micro': 'Moradia', 'valor': gasto_total * 0.4})
        transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'S', 'macro': 'Flexivel', 'micro': 'Restaurante/Delivery', 'valor': gasto_total * 0.25})
        transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'S', 'macro': 'Superfluo', 'micro': 'Lazer/Assinaturas', 'valor': gasto_total * 0.2})
        transacoes.append({'id_usuario': u, 'mes': mes, 'tipo': 'S', 'macro': 'Banco', 'micro': 'Juros/Tarifas', 'valor': gasto_total * 0.15})

df = pd.DataFrame(transacoes)
df['valor_direcional'] = np.where(df['tipo'] == 'E', df['valor'].abs(), -df['valor'].abs())

# ==============================================================================
# 2. CLUSTERIZAÇÃO DE COMPORTAMENTO E FAIXA SALARIAL (3 GRUPOS)
# ==============================================================================
df_mensal = df.groupby(['id_usuario', 'mes']).agg(
    entradas=('valor_direcional', lambda x: x[x>0].sum()),
    saidas=('valor_direcional', lambda x: x[x<0].abs().sum())
).reset_index()

df_mensal['saldo'] = df_mensal['entradas'] - df_mensal['saidas']
df_mensal['pct_sobra'] = np.where(df_mensal['entradas'] > 0, (df_mensal['saldo'] / df_mensal['entradas']) * 100, 0)

# Agrupando o perfil médio do ano (Corrigido para apenas numéricos)
perfil = df_mensal.groupby('id_usuario').mean(numeric_only=True).reset_index()

def classificar_grupo(row):
    if row['saldo'] < 0: return '1. Endividado (< 0)'
    elif row['pct_sobra'] < 15: return '2. Vulnerável (< 15%)'
    else: return '3. Saudável (>= 15%)'

perfil['cluster_financeiro'] = perfil.apply(classificar_grupo, axis=1)

def classificar_faixa(renda):
    if renda <= 2500: return 'Até R$ 2.500'
    elif renda <= 5000: return 'R$ 2.501 a R$ 5.000'
    elif renda <= 10000: return 'R$ 5.001 a R$ 10.000'
    else: return 'Acima de R$ 10.000'

perfil['faixa_salarial'] = perfil['entradas'].apply(classificar_faixa)

# Mapeamento do tipo de trabalhador
df_renda = df[df['tipo'] == 'E'].groupby(['id_usuario', 'micro'])['valor'].sum().unstack(fill_value=0)
perfil['perfil_trabalho'] = np.where(df_renda.get('Salario CLT', 0) > 0, 'SÓ CLT', 'SÓ AUTÔNOMO')

# ==============================================================================
# 3. PRODUTOS DO BANCO E MONETIZAÇÃO
# ==============================================================================
print("\n--- MATRIZ DE NEGÓCIOS ITAÚ ---")
gastos_banco = df[(df['macro'] == 'Banco') & (df['tipo'] == 'S')].groupby('id_usuario')['valor'].mean().reset_index()
gastos_banco.rename(columns={'valor': 'gasto_medio_banco'}, inplace=True)
perfil = pd.merge(perfil, gastos_banco, on='id_usuario')
perfil['pct_gasto_banco'] = (perfil['gasto_medio_banco'] / perfil['entradas']) * 100

def recomendar_produto(cluster):
    if 'Endividado' in cluster: return 'Renegociação IA / Crédito Consolidado'
    elif 'Vulnerável' in cluster: return 'Reserva Automática / Seguro Conta'
    else: return 'CDB Itaú Diário / Fundos Multimercado'

perfil['melhor_produto_itau'] = perfil['cluster_financeiro'].apply(recomendar_produto)
resumo_banco = perfil.groupby('cluster_financeiro').agg(
    PCT_Gasto_Banco=('pct_gasto_banco', 'mean'),
    Produto_Recomendado=('melhor_produto_itau', 'first')
).reset_index()
display(resumo_banco.round(2))

# ==============================================================================
# 4. COORTE DE DEZEMBRO: DÍVIDA ATUAL E MITIGAÇÃO
# ==============================================================================
df_dez = df_mensal[df_mensal['mes'] == 12]
negativados_dez = df_dez[df_dez['saldo'] < 0][['id_usuario', 'saldo']]
negativados_dez.rename(columns={'saldo': 'divida_atual'}, inplace=True)
negativados_dez['divida_atual'] = negativados_dez['divida_atual'].abs()

# Filtrando dados completos apenas dos negativados
ids_alvo = negativados_dez['id_usuario'].unique()
df_coorte = df[df['id_usuario'].isin(ids_alvo)]

# NOVAS VARIÁVEIS DE INTELIGÊNCIA DO AGENTE
# Score_de_Flexibilidade = % do orçamento gasto com coisas não-essenciais (Flexível + Supérfluo)
gastos_coorte = df_coorte[(df_coorte['mes'] == 12) & (df_coorte['tipo'] == 'S')].groupby(['id_usuario', 'macro'])['valor'].sum().unstack(fill_value=0).reset_index()
gastos_coorte['Total_Gastos'] = gastos_coorte[['Casa/Essencial', 'Flexivel', 'Superfluo', 'Banco']].sum(axis=1)

gastos_coorte['Score_de_Flexibilidade_Pct'] = ((gastos_coorte['Flexivel'] + gastos_coorte['Superfluo']) / gastos_coorte['Total_Gastos']) * 100
gastos_coorte = pd.merge(gastos_coorte, negativados_dez, on='id_usuario')

# Meta de Economia Mensal = Dividir a dívida em 6 meses para não estrangular o cliente
gastos_coorte['Meta_de_Economia_Mensal'] = gastos_coorte['divida_atual'] / 6

print("\n--- NOVAS VARIÁVEIS DO AGENTE (AMOSTRA) ---")
display(gastos_coorte[['id_usuario', 'divida_atual', 'Score_de_Flexibilidade_Pct', 'Meta_de_Economia_Mensal']].head().round(2))

# ==============================================================================
# 5. VISUALIZAÇÕES (DASHBOARD)
# ==============================================================================
sns.set_theme(style="whitegrid")
fig = plt.figure(figsize=(18, 12))

# CORREÇÃO DA SINTAXE DE GRID DO MATPLOTLIB
gs = plt.GridSpec(2, 2, figure=fig)

# Gráfico 1: Cluster de Clientes
ax1 = fig.add_subplot(gs[0, 0])
perc_cluster = perfil['cluster_financeiro'].value_counts(normalize=True).sort_index() * 100
sns.barplot(y=perc_cluster.index, x=perc_cluster.values, palette=['#d62728', '#ec7000', '#0046b5'], ax=ax1)
ax1.set_title('Status da População: Endividado, Vulnerável ou Saudável?', fontsize=12, fontweight='bold')
ax1.set_xlabel('Percentual da Base (%)')
for p in ax1.patches: ax1.text(p.get_width() + 1, p.get_y() + 0.4, f'{p.get_width():.1f}%', fontweight='bold')

# Gráfico 2: Autônomos vs CLT por Faixa Salarial
ax2 = fig.add_subplot(gs[0, 1])
ordem_faixas = ['Até R$ 2.500', 'R$ 2.501 a R$ 5.000', 'R$ 5.001 a R$ 10.000', 'Acima de R$ 10.000']
sns.countplot(data=perfil, x='faixa_salarial', hue='perfil_trabalho', order=ordem_faixas, palette={'SÓ CLT': '#0046b5', 'SÓ AUTÔNOMO': '#ec7000'}, ax=ax2)
ax2.set_title('Distribuição de Renda: Exclusivos CLT vs Exclusivos Autônomos', fontsize=12, fontweight='bold')
ax2.set_ylabel('Número de Clientes')
ax2.set_xlabel('')

# Gráfico 3: Coorte de Dezembro (Causa Raiz)
ax3 = fig.add_subplot(gs[1, :])

# CORREÇÃO PRINCIPAL: Remover colunas de string antes de tentar calcular a média
evolucao = df_mensal[df_mensal['id_usuario'].isin(ids_alvo)].drop(columns=['id_usuario']).groupby('mes').mean().reset_index()

meses_nome = {1:'Jan', 2:'Fev', 3:'Mar', 4:'Abr', 5:'Mai', 6:'Jun', 7:'Jul', 8:'Ago', 9:'Set', 10:'Out', 11:'Nov', 12:'Dez'}
evolucao['nome_mes'] = evolucao['mes'].map(meses_nome)

ax3.plot(evolucao['nome_mes'], evolucao['entradas'], label='Renda (Entradas)', color='#0046b5', marker='o', linewidth=3)
ax3.plot(evolucao['nome_mes'], evolucao['saidas'], label='Gastos (Saídas)', color='#d62728', marker='s', linewidth=3)
ax3.fill_between(evolucao['nome_mes'], evolucao['entradas'], evolucao['saidas'], where=(evolucao['saidas'] > evolucao['entradas']), interpolate=True, color='#d62728', alpha=0.2, label='Zona de Dívida')

ax3.set_title('O Caminho da Dívida: População Negativada no Fim do Ano', fontsize=14, fontweight='bold')
ax3.set_ylabel('R$ Médio')
ax3.legend()

plt.tight_layout()
plt.show()
```

O agente i.agora inverte essa lógica ao sugerir a *Renegociação IA / Crédito Consolidado* para o endividado e a *Reserva Automática* para o vulnerável, gerando valor sustentável de longo prazo para o ecossistema.

* **Interpretação:** Como quase metade do orçamento problemático está em categorias flexíveis, a **Meta de Economia Mensal** (repartir o buraco em parcelas suaves de ~R$ 115.00 a R$ 460.00) viabiliza a recuperação do cliente em 6 meses **sem tocar nas categorias essenciais** (moradia e saúde).

```python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore")

# ==============================================================================
# PRÉ-REQUISITO: GERAÇÃO DA BASE FACTUAL
# ==============================================================================
print("Carregando e processando a base factual de transações...")
np.random.seed(42)
n_clientes = 2500
usuarios = [f"CLT_{i:04d}" for i in range(n_clientes)]

# Criação de perfis variados de propensão ao gasto para gerar todos os 3 clusters
# 12% propensos a endividamento (>100%), 73% vulneráveis (85-98%), 15% saudáveis (<85%)
perfis_consumo = {
    u: np.random.choice([0.78, 0.93, 1.12], p=[0.15, 0.73, 0.12])
    for u in usuarios
}

transacoes = []
for mes in range(1, 13):
  for u in usuarios:
    is_clt = int(u.split("_")[1]) % 2 == 0
    renda = (
        np.random.normal(4500, 800) if is_clt else np.random.normal(3800, 1500)
    )
    renda = max(renda, 1000)

    # Aplicação do fator individual do perfil de consumo do correntista
    fator_u = perfis_consumo[u] * np.random.uniform(0.95, 1.05)

    gasto_casa = renda * 0.40 * fator_u
    gasto_flexivel = renda * np.random.uniform(0.30, 0.42) * fator_u
    gasto_superfluo = renda * 0.10 * fator_u
    gasto_banco = renda * 0.05 * fator_u

    transacoes.append({
        "id_usuario": u,
        "mes": mes,
        "tipo": "E",
        "macro": "Renda",
        "valor": renda,
    })
    transacoes.append({
        "id_usuario": u,
        "mes": mes,
        "tipo": "S",
        "macro": "Casa/Essencial",
        "valor": gasto_casa,
    })
    transacoes.append({
        "id_usuario": u,
        "mes": mes,
        "tipo": "S",
        "macro": "Flexivel/Delivery",
        "valor": gasto_flexivel,
    })
    transacoes.append({
        "id_usuario": u,
        "mes": mes,
        "tipo": "S",
        "macro": "Superfluo",
        "valor": gasto_superfluo,
    })
    transacoes.append({
        "id_usuario": u,
        "mes": mes,
        "tipo": "S",
        "macro": "Banco/Taxas",
        "valor": gasto_banco,
    })

df = pd.DataFrame(transacoes)
df["valor_direcional"] = np.where(df["tipo"] == "E", df["valor"], -df["valor"])

# Agrupamento mensal por cliente
df_mensal = (
    df.groupby(["id_usuario", "mes"])
    .agg(
        entradas=("valor_direcional", lambda x: x[x > 0].sum()),
        saidas=("valor_direcional", lambda x: x[x < 0].abs().sum()),
    )
    .reset_index()
)

df_mensal["saldo"] = df_mensal["entradas"] - df_mensal["saidas"]
df_mensal["pct_sobra"] = np.where(
    df_mensal["entradas"] > 0,
    (df_mensal["saldo"] / df_mensal["entradas"]) * 100,
    0,
)

perfil = df_mensal.groupby("id_usuario").mean(numeric_only=True).reset_index()


def classificar_grupo(row):
  if row["saldo"] < 0:
    return "1. Endividado (< 0)"
  elif row["pct_sobra"] < 15:
    return "2. Vulnerável (< 15%)"
  else:
    return "3. Saudável (>= 15%)"


perfil["cluster_financeiro"] = perfil.apply(classificar_grupo, axis=1)

# ==============================================================================
# PASSO 1: EXTRAÇÃO FACTUAL E VISUALIZAÇÃO GRÁFICA
# ==============================================================================
categorias_ordem = [
    "1. Endividado (< 0)",
    "2. Vulnerável (< 15%)",
    "3. Saudável (>= 15%)",
]
resumo_clusters = (
    perfil["cluster_financeiro"]
    .value_counts(normalize=True)
    .reindex(categorias_ordem, fill_value=0)
    * 100
)

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

cores_map = {
    "1. Endividado (< 0)": "#d62728",
    "2. Vulnerável (< 15%)": "#ec7000",
    "3. Saudável (>= 15%)": "#0046b5",
}

# Gráfico 1.1: Distribuição dos Clusters
sns.barplot(
    y=resumo_clusters.index,
    x=resumo_clusters.values,
    hue=resumo_clusters.index,
    palette=cores_map,
    legend=False,
    ax=axes[0],
)
axes[0].set_title(
    "Passo 1.1: Distribuição da Saúde Financeira dos Correntistas",
    fontsize=13,
    fontweight="bold",
)
axes[0].set_xlabel("Percentual da Base (%)", fontsize=11)
axes[0].set_ylabel("")
axes[0].set_xlim(0, max(resumo_clusters.values) + 15)

for p in axes[0].patches:
  width = p.get_width()
  if width > 0:
    axes[0].text(
        width + 1.0,
        p.get_y() + p.get_height() / 2,
        f"{width:.1f}%",
        fontweight="bold",
        fontsize=11,
        va="center",
    )

# Gráfico 1.2: Concentração de Gastos
gastos_macro = (
    df[df["tipo"] == "S"].groupby("macro")["valor"].sum().reset_index()
)
gastos_macro["pct"] = (gastos_macro["valor"] / gastos_macro["valor"].sum()) * 100
gastos_macro = gastos_macro.sort_values(by="pct", ascending=False)

sns.barplot(
    data=gastos_macro,
    x="pct",
    y="macro",
    hue="macro",
    palette="Blues_r",
    legend=False,
    ax=axes[1],
)
axes[1].set_title(
    "Passo 1.2: Comportamento Crítico (Concentração de Gastos)",
    fontsize=13,
    fontweight="bold",
)
axes[1].set_xlabel("Percentual do Orçamento Total (%)", fontsize=11)
axes[1].set_ylabel("")
axes[1].set_xlim(0, max(gastos_macro["pct"]) + 10)

for p in axes[1].patches:
  width = p.get_width()
  if width > 0:
    axes[1].text(
        width + 0.8,
        p.get_y() + p.get_height() / 2,
        f"{width:.1f}%",
        fontweight="bold",
        fontsize=11,
        va="center",
    )

plt.tight_layout()
plt.show()
```
