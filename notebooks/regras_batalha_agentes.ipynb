## Notebook de Regras, Recorte de Clientes e Score de Contexto

Este notebook contém a estrutura base para experimentação e modelagem das regras comportamentais dos 1.000 clientes, cálculo do Ponto de Índice e contextualização do Agente de Prompt.

```python
import pandas as pd
import numpy as np
import json

# 1. Simulação dos 1.000 clientes determinísticos
clientes = []
for cid in range(1, 1001):
    genero = 'F' if (cid % 2 == 0) else 'M'
    score = min(1000, max(120, ((cid * 37) % 850) + 150))
    
    if score >= 750:
        corte = 'ALTA_PROPENSAO'
    elif score >= 450:
        corte = 'MEDIA_PROPENSAO'
    else:
        corte = 'BAIXA_PROPENSAO'
        
    clientes.append({
        'id': cid,
        'genero': genero,
        'score_comportamental': score,
        'indice_corte': corte,
    })

df_clientes = pd.DataFrame(clientes)
print(f"Total de clientes simulados: {len(df_clientes)}")
print(df_clientes['genero'].value_counts())
df_clientes.head()
```
